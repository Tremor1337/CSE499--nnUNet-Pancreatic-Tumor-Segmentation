```python
import torch
from batchgenerators.utilities.file_and_folder_operations import join, maybe_mkdir_p
from nnunetv2.inference.predict_from_raw_data import nnUNetPredictor
from nnunetv2.paths import nnUNet_results, nnUNet_raw

# ---------- Basic config ----------
dataset_id = '110'
dataset_name = 'Dataset110_PANTHER_T1'
trainer = 'nnUNetTrainer__nnUNetPlans__3d_fullres'
folds = [0, 1, 2]
checkpoint = 'checkpoint_best.pth'

# ---------- Directories ----------
input_dir = join(nnUNet_raw, dataset_name, 'imagesTr')

# ---------- Run inference for each fold ----------
for fold in folds:
    print(f"\nRunning inference for Fold {fold}...")
    
    # Init predictor
    predictor = nnUNetPredictor(
        tile_step_size=0.5,
        use_gaussian=True,
        use_mirroring=True,
        perform_everything_on_device=True,
        device=torch.device('cuda', 0),
        verbose=True,
        verbose_preprocessing=False,
        allow_tqdm=True
    )

    predictor.initialize_from_trained_model_folder(
        join(nnUNet_results, f'{dataset_name}/{trainer}'),
        use_folds=(fold,),
        checkpoint_name=checkpoint
    )

    # Prediction output
    output_dir = join(nnUNet_results, f'predictions_fold{fold}')
    maybe_mkdir_p(output_dir)

    predictor.predict_from_files(
        input_dir,
        output_dir,
        save_probabilities=False,
        overwrite=True,
        num_processes_preprocessing=2,
        num_processes_segmentation_export=2
    )

print("\n All folds prediction complete.")
```

```python
import os
import json
import numpy as np
from pathlib import Path
import SimpleITK as sitk
import pandas as pd

# ===================== CONFIG =====================
DATASET_NAME = 'Dataset110_PANTHER_T1'     # nnU-Net dataset folder name
PRED_FOLD_DIRS = ['predictions_fold0', 'predictions_fold1', 'predictions_fold2']
OUT_DIR_NAME = 'eval_reports'               # will be created under nnUNet_results
# ==================================================

# Resolve nnU-Net env paths
nnUNet_raw = os.environ.get('nnUNet_raw')
nnUNet_results = os.environ.get('nnUNet_results')
if not nnUNet_raw or not nnUNet_results:
    raise EnvironmentError("Please ensure nnUNet_raw and nnUNet_results are set in your environment.")

label_dir = Path(nnUNet_raw) / DATASET_NAME / 'labelsTr'
pred_dirs = [Path(nnUNet_results) / d for d in PRED_FOLD_DIRS]
out_dir = Path(nnUNet_results) / OUT_DIR_NAME
out_dir.mkdir(parents=True, exist_ok=True)

# ---------------- Load class mapping ----------------
# Robust to either:
#   {"labels":{"0":"background","1":"pancreas","2":"tumor"}}
# or {"labels":{"background":0,"pancreas":1,"tumor":2}}
with open(Path(nnUNet_raw) / DATASET_NAME / 'dataset.json', 'r') as f:
    ds = json.load(f)

labels_dict = ds['labels']

label_map = {}  # id -> name
for k, v in labels_dict.items():
    if isinstance(k, str) and k.isdigit():
        label_map[int(k)] = str(v)
    elif isinstance(k, int):
        label_map[int(k)] = str(v)
    else:
        # keys are names, values are ids (possibly strings)
        label_map[int(v)] = str(k)

# Exclude background (assume id 0 if present)
class_ids = [cid for cid in sorted(label_map) if cid != 0]
if not class_ids:
    raise ValueError("No foreground classes found. Check dataset.json 'labels'.")

print("Label map (id->name):", label_map)
print("Classes evaluated:", [(cid, label_map[cid]) for cid in class_ids])

# --------------- Helpers ---------------
def read_nii(path: Path):
    img = sitk.ReadImage(str(path))
    arr = sitk.GetArrayFromImage(img)  # z,y,x
    return arr, img

def resample_like(moving_img, fixed_img, interp=sitk.sitkNearestNeighbor, out_dtype=sitk.sitkUInt16):
    return sitk.Resample(moving_img, fixed_img, sitk.Transform(), interp, 0, out_dtype)

def binarize(arr, class_id):
    return (arr == class_id).astype(np.uint8)

def safe_div(n, d):
    return float(n) / float(d) if d != 0 else 0.0

def dice(tp, fp, fn):
    return safe_div(2 * tp, 2 * tp + fp + fn)

def iou(tp, fp, fn):
    return safe_div(tp, tp + fp + fn)

def confusion_binary(gt_bin, pr_bin):
    tp = int(np.sum((gt_bin == 1) & (pr_bin == 1)))
    fp = int(np.sum((gt_bin == 0) & (pr_bin == 1)))
    fn = int(np.sum((gt_bin == 1) & (pr_bin == 0)))
    return tp, fp, fn

def eval_case(gt_arr, pr_arr, case_id, fold_name):
    rows = []
    for cid in class_ids:
        gt_bin = binarize(gt_arr, cid)
        pr_bin = binarize(pr_arr, cid)
        tp, fp, fn = confusion_binary(gt_bin, pr_bin)
        rows.append({
            'case': case_id,
            'fold': fold_name,
            'class_id': cid,
            'class_name': label_map[cid],
            'dice': dice(tp, fp, fn),
            'iou': iou(tp, fp, fn),
        })
    return rows

# Collect all GT label files
label_files = sorted(label_dir.glob('*.nii.gz'))
if not label_files:
    raise FileNotFoundError(f"No label files found in {label_dir}")

# --------------- Evaluate folds ---------------
all_rows = []
for pred_dir in pred_dirs:
    fold_name = pred_dir.name
    fold_rows = []
    if not pred_dir.exists():
        print(f"[WARN] Prediction directory missing: {pred_dir}")
        continue

    for gt_path in label_files:
        case_name = gt_path.name               # e.g., 10000.nii.gz
        pred_path = pred_dir / case_name       # prediction files must match GT names
        if not pred_path.exists():
            print(f"[WARN] Missing prediction for {case_name} in {fold_name}")
            continue

        gt_arr, gt_img = read_nii(gt_path)
        pr_arr, pr_img = read_nii(pred_path)

        # Geometry safety: resample pred to GT geometry if needed
        if (pr_img.GetSpacing()   != gt_img.GetSpacing() or
            pr_img.GetSize()      != gt_img.GetSize() or
            pr_img.GetDirection() != gt_img.GetDirection() or
            pr_img.GetOrigin()    != gt_img.GetOrigin()):
            pr_img = resample_like(pr_img, gt_img, interp=sitk.sitkNearestNeighbor, out_dtype=sitk.sitkUInt16)
            pr_arr = sitk.GetArrayFromImage(pr_img)

        fold_rows.extend(eval_case(gt_arr, pr_arr, case_name, fold_name))

    # Save per-fold CSV
    if fold_rows:
        df_fold = pd.DataFrame(fold_rows)
        df_fold.to_csv(out_dir / f'eval_{fold_name}.csv', index=False)
        all_rows.extend(fold_rows)
        print(f"[OK] Saved per-case metrics for {fold_name} -> {out_dir / f'eval_{fold_name}.csv'}")
    else:
        print(f"[WARN] No evaluable cases for {fold_name}")

# --------------- Summaries ---------------
if not all_rows:
    raise SystemExit("No results to summarize. Check your prediction folders & filenames.")

df = pd.DataFrame(all_rows)

# Per-fold, per-class mean/std
per_fold = (df.groupby(['fold','class_id','class_name'])
              .agg(dice_mean=('dice','mean'),
                   dice_std=('dice','std'),
                   iou_mean=('iou','mean'),
                   iou_std=('iou','std'))
              .reset_index())
per_fold.to_csv(out_dir / 'summary_per_fold.csv', index=False)

# Pooled across folds, per-class mean/std
overall = (df.groupby(['class_id','class_name'])
             .agg(dice_mean=('dice','mean'),
                  dice_std=('dice','std'),
                  iou_mean=('iou','mean'),
                  iou_std=('iou','std'))
             .reset_index())
overall.to_csv(out_dir / 'summary_overall.csv', index=False)

# Optional: macro averages per fold (across classes)
macro_per_fold = (df.groupby(['fold','case'])
                    .agg(dice_macro=('dice','mean'),
                         iou_macro=('iou','mean'))
                    .reset_index())
macro_per_fold.to_csv(out_dir / 'macro_by_case_per_fold.csv', index=False)

print("✅ Done.")
print(f"Per-fold CSVs: {', '.join([str(out_dir / f'eval_{p.name}.csv') for p in pred_dirs if (out_dir / f'eval_{p.name}.csv').exists()])}")
print(f"Summaries: {out_dir / 'summary_per_fold.csv'}, {out_dir / 'summary_overall.csv'}")
print(f"Macro by case: {out_dir / 'macro_by_case_per_fold.csv'}")
```

```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import SimpleITK as sitk

# ===================== CONFIG =====================
DATASET_NAME = "Dataset110_PANTHER_T1"
FOLD = "predictions_fold2"
# Use raw strings or forward slashes on Windows
NNUNET_RAW = r"E:\nnUNet_raw"
NNUNET_RESULTS = r"E:\nnUNet_results"
EVAL_CSV = os.path.join(NNUNET_RESULTS, "eval_reports", f"eval_{FOLD}.csv")

CLASS_TUMOR = 1        # visualize & rank by tumor
N_BEST = 10             # how many best cases to show
FIGSIZE = (12, 8)      # per-plane figure size
# ==================================================

# ---------- paths ----------
img_dir   = os.path.join(NNUNET_RAW, DATASET_NAME, "imagesTr")
label_dir = os.path.join(NNUNET_RAW, DATASET_NAME, "labelsTr")
pred_dir  = os.path.join(NNUNET_RESULTS, FOLD)

# ---------- utils ----------
def read_nii(path):
    img = sitk.ReadImage(path)
    arr = sitk.GetArrayFromImage(img)  # z,y,x
    return arr, img

def dice_score(a, b):
    a = a.astype(bool); b = b.astype(bool)
    inter = (a & b).sum()
    denom = a.sum() + b.sum()
    return (2.0 * inter / denom) if denom > 0 else 0.0

def iou_score(a, b):
    a = a.astype(bool); b = b.astype(bool)
    union = (a | b).sum()
    inter = (a & b).sum()
    return (inter / union) if union > 0 else 0.0

def colorize_label(lbl2d):
    """
    Map {0: background, 1: tumor, 2: pancreas} to RGB.
    background: dark blue, tumor: orange/red, tumor: green.
    """
    h, w = lbl2d.shape
    rgb = np.zeros((h, w, 3), dtype=np.float32)
    rgb[lbl2d == 0] = (0.05, 0.05, 0.25)   # background
    rgb[lbl2d == 1] = (0.85, 0.35, 0.15)   # tumor
    rgb[lbl2d == 2] = (0.10, 0.80, 0.30)   # pancreas
    return rgb

def pick_informative_slice(gt3d):
    """
    Return indices (axial_z, sagittal_x, coronal_y) where the GT foreground
    has the largest area in that plane. Falls back to center if empty.
    """
    fg = (gt3d > 0).astype(np.uint8)
    z_sums = fg.reshape(fg.shape[0], -1).sum(1)
    y_sums = fg.transpose(1,0,2).reshape(fg.shape[1], -1).sum(1)  # coronal along y
    x_sums = fg.transpose(2,0,1).reshape(fg.shape[2], -1).sum(1)  # sagittal along x
    z = int(np.argmax(z_sums)) if z_sums.sum() > 0 else fg.shape[0] // 2
    y = int(np.argmax(y_sums)) if y_sums.sum() > 0 else fg.shape[1] // 2
    x = int(np.argmax(x_sums)) if x_sums.sum() > 0 else fg.shape[2] // 2
    return z, x, y

def outline_from_mask(mask2d):
    """Return a thin outline (edges) of a binary mask."""
    from scipy.ndimage import binary_erosion
    mask = mask2d.astype(bool)
    er = binary_erosion(mask)
    outline = mask ^ er
    return outline.astype(np.uint8)

def panel_2x3_for_plane(ax_titles, img2d, gt2d, pred2d, dice_val, iou_val, plane_title):
    """
    ax_titles: ["Input", "Ground Truth", "Prediction"]
    """
    fig, axes = plt.subplots(2, 3, figsize=FIGSIZE)

    # Row 1: input | GT | pred
    axes[0,0].imshow(img2d, cmap='gray'); axes[0,0].set_title(ax_titles[0])
    axes[0,1].imshow(colorize_label(gt2d)); axes[0,1].set_title(ax_titles[1])
    axes[0,2].imshow(colorize_label(pred2d))
    axes[0,2].set_title(ax_titles[2] + f"\nDice: {dice_val:.3f}, IoU: {iou_val:.3f}")

    # Row 2: overlays + error map
    axes[1,0].imshow(img2d, cmap='gray')
    axes[1,0].imshow(colorize_label(gt2d), alpha=0.45)
    axes[1,0].set_title("Overlay GT")

    axes[1,1].imshow(img2d, cmap='gray')
    axes[1,1].imshow(colorize_label(pred2d), alpha=0.45)
    axes[1,1].set_title("Overlay Prediction")

    # Error map: FP/FN outlines on the image (for tumor class)
    gt_bin   = (gt2d == CLASS_TUMOR).astype(np.uint8)
    pred_bin = (pred2d == CLASS_TUMOR).astype(np.uint8)
    fn = ((gt_bin == 1) & (pred_bin == 0)).astype(np.uint8)
    fp = ((gt_bin == 0) & (pred_bin == 1)).astype(np.uint8)
    fn_edge = outline_from_mask(fn)
    fp_edge = outline_from_mask(fp)

    axes[1,2].imshow(img2d, cmap='gray')
    axes[1,2].imshow(np.ma.masked_where(fn_edge == 0, fn_edge), alpha=0.9, cmap='gray')   # white-ish
    axes[1,2].imshow(np.ma.masked_where(fp_edge == 0, fp_edge*0), alpha=0.9, cmap='gray') # dark outline
    axes[1,2].set_title("Error Map")

    for ax in axes.ravel():
        ax.axis('off')

    plt.suptitle(plane_title, y=0.98)
    plt.tight_layout()
    return fig, axes

def show_case_all_planes(case_name):
    img_p  = os.path.join(img_dir,   case_name.replace(".nii.gz", "_0000.nii.gz"))
    gt_p   = os.path.join(label_dir, case_name)
    pred_p = os.path.join(pred_dir,  case_name)

    img3d, _  = read_nii(img_p)
    gt3d, _   = read_nii(gt_p)
    pr3d, _   = read_nii(pred_p)

    # pick informative slices
    z_axial, x_sag, y_cor = pick_informative_slice(gt3d)

    # tumor only for metrics
    gt_tum_ax  = (gt3d[z_axial] == CLASS_TUMOR).astype(np.uint8)
    pr_tum_ax  = (pr3d[z_axial] == CLASS_TUMOR).astype(np.uint8)
    dice_ax    = dice_score(gt_tum_ax, pr_tum_ax)
    iou_ax     = iou_score(gt_tum_ax, pr_tum_ax)

    gt_tum_sag = (gt3d[:, :, x_sag] == CLASS_TUMOR).astype(np.uint8)
    pr_tum_sag = (pr3d[:, :, x_sag] == CLASS_TUMOR).astype(np.uint8)
    dice_sag   = dice_score(gt_tum_sag, pr_tum_sag)
    iou_sag    = iou_score(gt_tum_sag, pr_tum_sag)

    gt_tum_cor = (gt3d[:, y_cor, :] == CLASS_TUMOR).astype(np.uint8)
    pr_tum_cor = (pr3d[:, y_cor, :] == CLASS_TUMOR).astype(np.uint8)
    dice_cor   = dice_score(gt_tum_cor, pr_tum_cor)
    iou_cor    = iou_score(gt_tum_cor, pr_tum_cor)

    # AXIAL panel
    panel_2x3_for_plane(
        ["Input", "Ground Truth", "Prediction"],
        img3d[z_axial], gt3d[z_axial], pr3d[z_axial],
        dice_ax, iou_ax, f"Axial Plane (Slice {z_axial})"
    )
    # SAGITTAL panel
    panel_2x3_for_plane(
        ["Input", "Ground Truth", "Prediction"],
        img3d[:, :, x_sag], gt3d[:, :, x_sag], pr3d[:, :, x_sag],
        dice_sag, iou_sag, f"Sagittal Plane (Slice {x_sag})"
    )
    # CORONAL panel
    panel_2x3_for_plane(
        ["Input", "Ground Truth", "Prediction"],
        img3d[:, y_cor, :], gt3d[:, y_cor, :], pr3d[:, y_cor, :],
        dice_cor, iou_cor, f"Coronal Plane (Slice {y_cor})"
    )
    plt.show()

# ---------- pick best only ----------
eval_df = pd.read_csv(EVAL_CSV)
tumor_rows = eval_df[eval_df["class_id"] == CLASS_TUMOR]

# mean tumor dice per case (fold2 file already per-case; group just in case)
case_scores = tumor_rows.groupby("case")["dice"].mean().reset_index()

# sort high->low and trim to N_BEST
best_cases = case_scores.sort_values("dice", ascending=False)["case"].tolist()
cases_to_show = best_cases[:N_BEST]

print("Best cases:", cases_to_show)

# ---------- render ----------
for case in cases_to_show:
    print(f"\n=== {case} ===")
    show_case_all_planes(case)
```
